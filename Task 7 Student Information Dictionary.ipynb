# Student Information Dictionary
**Objective:** Practice dictionaries and structured data representation by creating a dictionary containing student information and implementing operations to search, add, and update student records.

```python
# Initialize the student dictionary using unique student IDs as keys
students_db = {
    'S101': {'name': 'Alice Smith', 'age': 20, 'major': 'Computer Science'},
    'S102': {'name': 'Bob Jones', 'age': 22, 'major': 'Mathematics'},
    'S103': {'name': 'Charlie Brown', 'age': 19, 'major': 'Physics'}
}

print("Initial student database loaded successfully.")
```

## 1. Add Operation
Function to add a new student record to the dictionary. It checks if the unique student ID already exists to prevent overwriting.

```python
def add_student(student_id, name, age, major):
    """Adds a new student to the dictionary."""
    if student_id in students_db:
        print(f"Error: Student ID '{student_id}' already exists.")
    else:
        students_db[student_id] = {'name': name, 'age': age, 'major': major}
        print(f"Success: Student '{name}' added with ID '{student_id}'.")
```

## 2. Update Operation
Function to update an existing student's information. It allows partial updates (e.g., updating just the age without changing the major).

```python
def update_student(student_id, name=None, age=None, major=None):
    """Updates an existing student's record."""
    if student_id in students_db:
        if name:
            students_db[student_id]['name'] = name
        if age:
            students_db[student_id]['age'] = age
        if major:
            students_db[student_id]['major'] = major
        print(f"Success: Student ID '{student_id}' has been updated.")
    else:
        print(f"Error: Student ID '{student_id}' not found.")
```

## 3. Search Operation
Function to search for a student by their unique ID and display their information.

```python
def search_student(student_id):
    """Searches for a student by ID and prints their details."""
    if student_id in students_db:
        info = students_db[student_id]
        print(f"--- Student Record: {student_id} ---")
        print(f"Name:  {info['name']}")
        print(f"Age:   {info['age']}")
        print(f"Major: {info['major']}")
        print("------------------------------")
    else:
        print(f"Search Failed: Student ID '{student_id}' not found.")
```

## 4. Example Outputs
Testing the add, update, and search functions.

```python
# Test 1: Add a new student
print("--- Testing Add Operation ---")
add_student('S104', 'Diana Prince', 21, 'History')

# Test 2: Attempt to add an existing student
add_student('S101', 'Duplicate Alice', 99, 'None')

# Test 3: Update an existing student
print("\n--- Testing Update Operation ---")
update_student('S102', age=23, major='Applied Mathematics')

# Test 4: Search for updated and new students
print("\n--- Testing Search Operation ---")
search_student('S102')
search_student('S104')
search_student('S999') # Non-existent ID
```

## Interview Questions

* **What is a dictionary in Python?**
  A dictionary is a built-in Python data structure that stores data in key-value pairs. It is mutable, unordered (though insertion order is preserved in modern Python versions), and indexed by keys.
  
* **Why are keys important in dictionaries?**
  Keys act as unique identifiers for the values stored in a dictionary. They allow for rapid, highly efficient data retrieval, updating, and deletion without needing to iterate through the entire structure.
  
* **How is a dictionary different from a list?**
  Lists are ordered collections of items accessed by their numerical index (e.g., `list[0]`). Dictionaries are collections of key-value pairs accessed by unique keys (e.g., `dict['student_id']`). Lists are ideal for ordered sequences, while dictionaries are ideal for structured associations and fast lookups.
